```python
# Exercício 1: Peça ao usuário para digitar vários números inteiros (um por vez).
# Quando ele digitar 0, pare de pedir números.
# Armazene todos os números em uma lista e depois:
# - Imprima a média dos números
# - Imprima o maior e o menor valor
numeros = []
while True:
    num = input('Digite um número inteiro ou 0 para finalizar.')
    while not num.isnumeric():
        num = input('Digite um número inteiro ou 0 para finalizar.')
    num = int(num)
    if num == 0:
        break
    numeros.append(num)
soma = 0
maior = numeros[0]
menor = numeros[0]
for i in numeros:
    soma += i
    if maior < i:
        maior = i
    elif menor > i:
        menor = i
media = soma / len(numeros)
print(media)
print(f'Maior valor da lista: {maior}')
print(f'Menor valor da lista: {menor}')
```

```python
# Exercício 2: Crie uma lista com números de 1 a 20.
# Depois, remova todos os múltiplos de 3 dessa lista (sem usar outra lista).
numeros = list(range(1, 21))
i = 0
while i < len(numeros):
    if numeros[i] % 3 == 0:
        del numeros[i]
    else:
        i += 1
print(numeros)
```
